```python

from bs4 import BeautifulSoup as bs
import urllib
import urllib.request
import re
import pandas as pd

r = urllib.request.urlopen("https://centrecolonels.com/sports/baseball/stats/2021/hendrix/boxscore/4638")

# Convert to a BS object
game_soup = bs(r, "html.parser")

# Print out html
#print(game_soup.prettify())

#pbp = game_soup.find('section', {'id':'play-by-play'})
#print(pbp)
k=1

table = game_soup.find_all('table', attrs={'class':'sidearm-table play-by-play'})
#print(table)

for period in table:
    if k == 1:
        headers=[]
        #total_headers = 0
        for i in period.find_all('thead'):
            for i in period.find_all('th'):
                title = i.text.strip()
                headers.append(title)
                #total_headers += 1
        #print(headers)
        #total_headers = (total_headers / 2) - 1
        #print(total_headers)
        headers = headers[0:3]
                
        df = pd.DataFrame(columns = headers)
    print(df)
    
    for row in period.find_all('tr')[1:]:
        data = row.find_all(['th', 'td'])
        #print(data)
        row_data = [td.text.strip() for td in data]
        print(row_data)
        length = len(df)
        df.loc[length] = row_data
        #print(row_data)

        # data = row.find_all('td')
        # row_data = [td.text.strip() for td in data]
        # length=len(df)
        # df.loc[length] = row_data
        


    
    
    k=k+1
    #df= pd.DataFrame()
#df.to_csv(r'/Users/ericmerdian/Desktop/Desktop/A_Summer_Research/field-hockey/test/export_dataframe.csv', index = False, header=False)
#print(df)
```
