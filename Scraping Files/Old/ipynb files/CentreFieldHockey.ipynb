## Task #1: Get info box for Toy Story 3

```python
r = requests.get("https://centrecolonels.com/sports/mens-lacrosse/stats/2021/piedmont/boxscore/4598")

# Convert to a BS object
soup = bs(r.content)

# Print out html
print(soup.prettify())
```

```python
table = soup.find('table', {'class':'sidearm-table play-by-play'})
```

```python
table
```

```python
headers = []

for i in table.find_all('thead'):
    for i in table.find_all('th'):
        title = i.text.strip()
        headers.append(title)
```

```python
headers = headers[:9]
```

```python
headers
```

```python
df = pd.DataFrame(columns = headers)
```

```python
df
```

```python
for row in table.find_all('tr')[1:]:
    data = row.find_all('td')
    row_data = [td.text.strip() for td in data]
    length = len(df)
    df.loc[length] = row_data
```
