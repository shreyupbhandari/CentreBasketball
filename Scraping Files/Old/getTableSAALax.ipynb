```python
import requests
from bs4 import BeautifulSoup as bs
import re
import pandas as pd
import fileinput
import os
import urllib
import urllib.request, urllib.error
import math
from pathlib import Path


def getHTML(sport, year1, school):
    print('start')
    sport_schedule = open(r'/Users/shelbyyoung/Desktop/sports analytics/SAA files/'+school +'/' + sport + '/' + str(year1) + '_links.csv', 'r')
    sport_games = [line.rstrip('\n') for line in sport_schedule]
    sport_schedule.close()
    #print(sport_games)

    for game_url in sport_games:
        print(game_url)
        list_game = game_url.split("/")
        try:
            game_opponent = list_game[7]
            game_id= list_game[9]
        except urllib.error.URLError:
            continue

        r = requests.get(game_url)

        soup = bs(r.content, features='lxml')

        try:
            k = 1
            table = soup.find_all('table', attrs = {'class':'sidearm-table play-by-play'})
            #print(table)
            innings = []
            three_headers = ['softball', 'baseball']
            no_innings = ['mens-soccer', 'womens-soccer', 'field-hockey']
            per_num = []
            for period in table:
                for i in period.find_all('caption'):
                    inning = i.text.strip()
                    innings.append(inning)
                if k == 1:
                    headers=[]
                            
                    for i in period.find_all('thead'):
                        for i in period.find_all('th'):
                            title = i.text.strip()
                            headers.append(title)

                    if sport in three_headers:
                        headers = headers[0:3]
                    else:
                        headers = headers[0:9]
                    df = pd.DataFrame(columns = headers)
                length=len(df)
                if sport in three_headers:
                    df.loc[length] = [inning] + [''] * 2
                elif sport not in no_innings:
                    df.loc[length] = ['Start Period ' + str(k)] + [''] * 8
                    per_num.append(k)
                

                for row in period.find_all('tr')[1:]:
                    j = 0
                    for data in row.find_all('td'):
                        j+= 1
                    if j == 5:
                        data = row.find_all('td', attrs = {'colspan': '5'})
                        row_data = [td.text.strip() for td in data]
                        length=len(df)
                        df.loc[length] = [''] * 3 + row_data + [''] * 5
                    else:
                        data = row.find_all(['td', 'th'])
                        row_data = [td.text.strip() for td in data]
                        length=len(df)
                        df.loc[length] = row_data
                        #print(k)

                    per_num.append(k)

                
                k=k+1
                
            if sport == 'field-hockey':
                home_team_score = 0
                visiting_team_score = 0
                for ind in df.index:
                    if df['Home Team Score'][ind] == '':
                        df['Home Team Score'][ind] = home_team_score
                    else:
                        home_team_score = df['Home Team Score'][ind]
                    if df['Visiting Team Score'][ind] == '':
                        df['Visiting Team Score'][ind] = visiting_team_score
                    else:
                        visiting_team_score = df['Visiting Team Score'][ind]
                        
            if (sport == 'womens-lacrosse') or (sport == 'mens-lacrosse'):
                visiting_header = df.columns[2]
                home_header = df.columns[4]
                home_team_score = 0
                visiting_team_score = 0
                for ind in df.index:
                    #print(df[home_header][ind])
                    if df[home_header][ind] == '':
                        df[home_header][ind] = home_team_score
                    else:
                        home_team_score = df[home_header][ind]
                    if df[visiting_header][ind] == '':
                        df[visiting_header][ind] = visiting_team_score
                    else:
                        visiting_team_score = df[visiting_header][ind]

            #print(df)
            #print(len(df))
            #print(len(per_num))
            df['period number'] = per_num
            print('made it')

            output_file = str(game_id) + '_' + game_opponent + '.csv'

            output_dir = Path('/Users/shelbyyoung/Desktop/sports analytics/SAA files/' + school +'/'+ sport + '/Tables/' + str(year1) + '/')

            output_dir.mkdir(parents=True, exist_ok=True)

            df.to_csv(output_dir / output_file)
            print('made table')
        except:
            continue

def main():
    
    lacrosse_schools = ['berryvikings', 'bscsports', 'hendrixwarriors', 'rhodeslynx', 'sewaneetigers']
    
    for school in lacrosse_schools: #change based on sport
        print('school: ', school)
        
        for year in range(2011,2022):
            getHTML('mens-lacrosse', year, school)

main()
```
